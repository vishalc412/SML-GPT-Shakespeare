```python
import os

all_texts = []
shakespeare_files = [
    "/content/alls-well-that-ends-well_TXT_FolgerShakespeare.txt",
    "/content/antony-and-cleopatra_TXT_FolgerShakespeare.txt",
    "/content/coriolanus_TXT_FolgerShakespeare.txt",
    "/content/as-you-like-it_TXT_FolgerShakespeare.txt",
    "/content/cymbeline_TXT_FolgerShakespeare.txt",
    "/content/hamlet_TXT_FolgerShakespeare.txt"
]

for file_path in shakespeare_files:
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            all_texts.append(f.read())
    except FileNotFoundError:
        print(f"Warning: File not found at {file_path}. Skipping.")

combined_text = "\n".join(all_texts)
print(f"Combined dataset length: {len(combined_text)} characters")
print(combined_text[:500]) # Display the first 500 characters to verify
```

```python
chars = sorted(list(set(combined_text)))
vocab_size = len(chars)
print('Total Characters:', len(combined_text))
print('Unique Characters:', vocab_size)
print(f"All characters : {''.join(chars)}")
```

```python
## Tokenication

stoi = {ch :i for i, ch in enumerate(chars)} #string to int
itos = {i :ch for i, ch in enumerate(chars)} #int to string

##encoder text --> numerics
encode = lambda s:[stoi[c] for c in s]

#decoder number to text
decode = lambda l: ''.join([itos[i] for i in l])

#test it
encode = encode('hello')
print(encode)
decode = decode(encode)
print(decode)
```
